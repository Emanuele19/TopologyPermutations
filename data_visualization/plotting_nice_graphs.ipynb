```python
import pandas as pd
import numpy as np
import ast
import plotly.express as px
import plotly.io as pio
from sklearn.decomposition import PCA

# Caricare il primo dataset (embedding)
file_path_embedding = "./pre_processing_output/duplicated_nodes_main_components.csv"  # Modifica con il percorso corretto
df = pd.read_csv(file_path_embedding)

# Convertire la colonna embedding in liste di numeri
df["GO_embeddings"] = df["GO_embeddings"].apply(ast.literal_eval)
embeddings_df = pd.DataFrame(df["GO_embeddings"].tolist())

# Caricare il secondo dataset (similarità)
file_path_similarity = "./pre_processing_output/duplicated_nodes_main_components_sim.csv"  # Modifica con il percorso corretto
df_sim = pd.read_csv(file_path_similarity)

# Convertire la colonna embedding in liste di numeri
df_sim["GO_embeddings"] = df_sim["GO_embeddings"].apply(ast.literal_eval)
similarity_df = pd.DataFrame(df_sim["GO_embeddings"].tolist())

# Applicare PCA con 3 componenti per visualizzazione 3D
pca_3d = PCA(n_components=3)

# Trasformazione per il primo dataset (embedding)
embedding_3D = pca_3d.fit_transform(embeddings_df)
df_pca_3d_embed = pd.DataFrame(embedding_3D, columns=["PC1", "PC2", "PC3"])
df_pca_3d_embed["label"] = df["label"]

# Trasformazione per il secondo dataset (similarità)
similarity_3D = pca_3d.fit_transform(similarity_df)
df_pca_3d_sim = pd.DataFrame(similarity_3D, columns=["PC1", "PC2", "PC3"])
df_pca_3d_sim["label"] = df_sim["label"]

# Creazione del plot 3D interattivo per gli embedding
fig_embed = px.scatter_3d(df_pca_3d_embed, x="PC1", y="PC2", z="PC3", 
                          color=df_pca_3d_embed["label"].astype(str), 
                          title="PCA 3D - Embedding",
                          labels={"color": "Label"},
                          opacity=0.7)

# Creazione del plot 3D interattivo per le similarità
fig_sim = px.scatter_3d(df_pca_3d_sim, x="PC1", y="PC2", z="PC3", 
                        color=df_pca_3d_sim["label"].astype(str), 
                        title="PCA 3D - Similarità",
                        labels={"color": "Label"},
                        opacity=0.7)

# Salvare i grafici in HTML per visualizzarli in locale
fig_embed.write_html("pca_3d_embedding.html")
fig_sim.write_html("pca_3d_similarity.html")

print("File salvati! Apri 'pca_3d_embedding.html' e 'pca_3d_similarity.html' nel browser per esplorarli.")
```
