```python
# export_anc2vec_embeddings_py36.py
import numpy as np
import anc2vec

# 1) prendi il dict {GO: vector}
emb = anc2vec.get_embeddings()  # dict: GO -> list/np.ndarray

# 2) ordina le chiavi per stabilità
terms = sorted(emb.keys())

# 3) costruisci matrice [N, D]
vectors = np.stack([np.asarray(emb[t], dtype=np.float32) for t in terms], axis=0)  # [N,D]
terms_arr = np.array(terms, dtype="U32")  # GO:0000001 ecc.

# 4) salva in formato portabile
np.savez_compressed("anc2vec_go_embeddings_v1.npz", terms=terms_arr, vectors=vectors)

print("Salvato: anc2vec_go_embeddings_v1.npz")
print("shape:", vectors.shape, "num_terms:", len(terms))
```
