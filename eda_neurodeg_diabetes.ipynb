```python
import pandas as pd

AD_PATH = './sources/UniProtKB-Alzheimer.tsv'
PD_PATH = './sources/UniProtKB-Parkinson.tsv'
D1_PATH = './sources/test_db/UniProtKB-MS.tsv'
D2_PATH = './sources/test_db/UniProtKB-SLE.tsv'

ad_csv = pd.read_csv(AD_PATH, sep='\t')
pd_csv = pd.read_csv(PD_PATH, sep='\t')
t1_csv = pd.read_csv(D1_PATH, sep='\t')
t2_csv = pd.read_csv(D2_PATH, sep='\t')

entry_set_ad = set(ad_csv['Entry'].to_list())
entry_set_pd = set(pd_csv['Entry'].to_list())
entry_set_t1 = set(t1_csv['Entry'].to_list())
entry_set_t2 = set(t2_csv['Entry'].to_list())

print('C1 len: ', len(entry_set_t1))
print('C2 len: ', len(entry_set_t2))

common_test = set.intersection(entry_set_t1, entry_set_t2)
print('Common test_db proteins', len(common_test), "\n", common_test)

entry_set_neuro = set.union(entry_set_pd, entry_set_ad)
entry_set_diabetes = set.union(entry_set_t1, entry_set_t2)

print("neurodegenerative proteins: ", len(entry_set_neuro))
print("diabetes proteins: ", len(entry_set_diabetes))

common = set.intersection(entry_set_diabetes, entry_set_neuro)
print("common proteins: ", len(common), "\n", common)

print("common of commons: ", len(set.intersection(common_test, common)), "\n", set.intersection(common_test, common))
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Funzione per pulire ed esplodere i GO term
def process_go_column(df, class_label):
    # Rimuoviamo eventuali ";" finali (visto il problema precedente!) e spazi
    df_clean = df[['Gene Ontology (GO)']].dropna().copy()
    df_clean['GO'] = df_clean['Gene Ontology (GO)'].str.rstrip(';').str.split(';')
    
    # Esplodiamo la lista in singole righe
    df_exploded = df_clean.explode('GO')
    
    # Pulizia da spazi bianchi extra e rimozione di eventuali stringhe vuote
    df_exploded['GO'] = df_exploded['GO'].str.strip()
    df_exploded = df_exploded[df_exploded['GO'] != '']
    
    # Aggiungiamo la colonna della classe per il confronto successivo
    df_exploded['Classe'] = class_label
    return df_exploded[['GO', 'Classe']]

# Elaboriamo i due dataset
go_t1 = process_go_column(t1, 'Classe 1')
go_t2 = process_go_column(t2, 'Classe 2')

# Uniamo i due dataset esplosi in un unico DF globale
df_all_go = pd.concat([go_t1, go_t2], ignore_index=True)
```

```python
# Calcoliamo il conteggio per ogni GO in ogni classe
go_counts = df_all_go.groupby(['Classe', 'GO']).size().reset_index(name='Conteggio')

# Calcoliamo il totale dei GO term per ogni classe per fare la percentuale
totale_per_classe = df_all_go.groupby('Classe').size()
go_counts['Percentuale'] = go_counts.apply(
    lambda row: (row['Conteggio'] / totale_per_classe[row['Classe']]) * 100, axis=1
)
```

```python
# Selezioniamo i 20 GO term più frequenti in assoluto per il grafico
top_20_go = df_all_go['GO'].value_counts().head(20).index
df_plot = go_counts[go_counts['GO'].isin(top_20_go)]

# Ordiniamo il plot per rendere il grafico più leggibile
df_plot = df_plot.sort_values(by='Percentuale', ascending=False)

# Creazione del grafico
plt.figure(figsize=(12, 8))
sns.barplot(
    data=df_plot, 
    y='GO', 
    x='Percentuale', 
    hue='Classe', 
    palette='Set2'
)

plt.title('Top 20 Gene Ontology (GO) Terms - Confronto Percentuale tra Classi')
plt.xlabel('% sul totale dei GO term della classe')
plt.ylabel('GO Term')
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Creiamo la pivot table partendo dalle percentuali calcolate prima
pivot_df = go_counts.pivot(index='GO', columns='Classe', values='Percentuale').fillna(0)

# 2. Calcoliamo il Log2 Fold Change
# Usiamo un epsilon per evitare il logaritmo di zero o divisioni per zero
epsilon = 0.001 
pivot_df['Log2_FC'] = np.log2((pivot_df['Classe 1'] + epsilon) / (pivot_df['Classe 2'] + epsilon))

# 3. Filtriamo per tenere i termini biologicamente rilevanti
# Escludiamo i termini troppo rari (es. che non raggiungono lo 0.5% in nessuna delle due classi)
soglia_presenza = 0.5
pivot_df_filtered = pivot_df[(pivot_df['Classe 1'] > soglia_presenza) | (pivot_df['Classe 2'] > soglia_presenza)].copy()

# 4. Selezioniamo i top sbilanciati da una parte e dall'altra
# Prendiamo ad esempio i 15 più arricchiti in Classe 1 e i 15 più arricchiti in Classe 2
top_classe1 = pivot_df_filtered.sort_values(by='Log2_FC', ascending=False).head(15)
top_classe2 = pivot_df_filtered.sort_values(by='Log2_FC', ascending=True).head(15)

# Uniamo i due estremi per il plot e ordiniamo dal più positivo al più negativo
df_plot_fc = pd.concat([top_classe1, top_classe2]).sort_values(by='Log2_FC', ascending=False).reset_index()

# 5. Creazione del grafico
plt.figure(figsize=(12, 10))

# Usiamo una mappa di colori divergente (es. Coolwarm) basata sul valore di Log2_FC
colors = ['#4C72B0' if x > 0 else '#C44E52' for x in df_plot_fc['Log2_FC']]

sns.barplot(
    data=df_plot_fc,
    x='Log2_FC',
    y='GO',
    palette=colors
)

# Linea di sbarramento sullo zero (nessuna differenza)
plt.axvline(x=0, color='black', linestyle='--', linewidth=1.2)

# Abbellimenti grafici e annotazioni simmetriche
plt.title('Differenza di Abbondanza dei GO Terms (Log2 Fold Change)', fontsize=14, pad=15)
plt.xlabel('← Più frequente in Classe 2      |      Più frequente in Classe 1 →\n[ Log2 ( % Classe 1 / % Classe 2 ) ]', fontsize=12)
plt.ylabel('Gene Ontology (GO) Term', fontsize=12)
plt.grid(axis='x', linestyle=':', alpha=0.6)

plt.tight_layout()
plt.show()
```
