```python
source_path = '../files/sources'
gene_diseases_path = f'{source_path}/CTD_curated_genes_diseases.csv'
disease_pathway_path = f'{source_path}/CTD_diseases_pathways.csv'
gobp_disease_path = f'{source_path}/CTD_Phenotype-Disease_biological_process_associations.tsv'
```

```python
import pandas as pd

column_names = ['GeneSymbol', 'GeneID', 'DiseaseName', 'DiseaseID', 'OmimIDs', 'PubMedIDs']

gene_diseases_df = pd.read_csv(gene_diseases_path, skiprows=29, names=column_names)

gene_diseases_df_AD = gene_diseases_df[gene_diseases_df["DiseaseName"].str.lower().str.startswith("alzheimer disease", na=False)].reset_index(drop=True)
gene_diseases_df_PD = gene_diseases_df[gene_diseases_df["DiseaseName"].str.lower().str.startswith("parkinson disease", na=False)].reset_index(drop=True)

gene_diseases_df_PD
```

```python
column_names = ['DiseaseName', 'DiseaseID', 'PathwayName', 'PathwayID', 'InferenceGeneSymbol']

disease_pathway_df = pd.read_csv(disease_pathway_path, skiprows=29, names=column_names)

disease_pathway_df_AD = disease_pathway_df[disease_pathway_df["DiseaseName"].str.lower().str.startswith("alzheimer disease", na=False)].reset_index(drop=True)
disease_pathway_df_PD = disease_pathway_df[disease_pathway_df["DiseaseName"].str.lower().str.startswith("parkinson disease", na=False)].reset_index(drop=True)

disease_pathway_df_AD
```

```python
column_names = ['GOName', 'GOID', 'DiseaseName', 'DiseaseID', 'InferenceChemicalQty', 'InferenceChemicalNames', 'InferenceGeneQty', 'InferenceGeneSymbols']

gobp_disease_df = pd.read_csv(gobp_disease_path, skiprows=29, names=column_names, sep='\t')

gobp_disease_df_AD = gobp_disease_df[gobp_disease_df["DiseaseName"].str.lower().str.startswith("alzheimer disease", na=False)].reset_index(drop=True)
gobp_disease_df_PD = gobp_disease_df[gobp_disease_df["DiseaseName"].str.lower().str.startswith("parkinson disease", na=False)].reset_index(drop=True)

gobp_disease_df_AD[~gobp_disease_df_AD['InferenceGeneSymbols'].isna()]

all_genes = gobp_disease_df_AD["InferenceGeneSymbols"].dropna().str.split("|")
gene_set = set(gene for sublist in all_genes for gene in sublist)

all_gos = gobp_disease_df_AD["GOID"].tolist()
go_set = set(all_gos)
```

```python
uniprot_go_path = f'{source_path}/UniProtKB-Alzheimer.tsv'
uniprot_go_df = pd.read_csv(uniprot_go_path, sep='\t')

all_uniprot_gos = uniprot_go_df['Gene Ontology IDs'].dropna().str.split(';')
uniprot_go_set = set(gene.strip() for sublist in all_uniprot_gos for gene in sublist)

print(f"Numero di GO in CDT: {len(go_set)}")
print(f"Numero di GO in UniProt: {len(uniprot_go_set)}")
print(f"Numero di GO non attinenti a AD: {len(uniprot_go_set - go_set)}")
print(f"Numero di GO in attinenti ad AD: {len(uniprot_go_set & go_set)}")

import pickle as pk

with open('CTD_gos.pkl', 'wb') as f:
    pk.dump(go_set, f)
```
