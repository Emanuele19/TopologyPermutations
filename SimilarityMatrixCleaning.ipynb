```python
import pandas as pd

# Funzione che prende come parametro il path della matrice di similarità e il nome dell'output
# Le pulizie che vengono effettuate sono:
# - Mapping NA -> 0
# - Arrotondamento a 3a cifra significativa
# - Toglie ultima riga e ultima colonna (NA)

def clean(src_path, dst_path):
    src = pd.read_csv(src_path, na_values=["NA"])
    # Eliminazione ultima riga e colonna
    src = src.iloc[:-1, :]
    src = src.iloc[:, :-1]

    # Mapping NA -> 0
    src.replace("NA", 0).fillna(0)

    # Arrotondamento
    src.round(3)

    # Replace NA -> 0
    src.to_csv(dst_path, na_rep="0", index=False)
    return pd.read_csv(dst_path)
```
